# Data Structure Guide

### Zip file
___

```
your_zip.zip
├── folder1
│   ├── audio.wav
│   └── audio.lab
│   └── ...
├── folder2
│   ├── audio.wav
│   └── audio.lab
│   └── ...
├── folder3
│   ├── audio.wav
│   └── audio.lab
│   └── ...
...
```

> Folder names will be used as language name. For language code, look at langs.txt you got after preprocessing

- Both `.wav` and `.lab` filenames should match within each folder (example: `song.wav` and `song.lab`)
___

### Example

```
(multi-lang)
data.zip
├── JPN
│   ├── sample_001.wav
│   └── sample_001.lab
│   └── ...
├── ENG
│   ├── sample_001.wav
│   └── sample_001.lab
│   └── ...
```

```
(single-lang)
data.zip
├── JPN
│   ├── sample_001.wav
│   └── sample_001.lab
│   └── ...
```

Please try to avoid bad labels like trash phonemes or labels, there's a risk of them getting predicted!

````python
#@title # Edit Config
# copied from @HAI-D, im too lazy to make this, ty <333

data_dir = "/content/training_dataset"
#@markdown # Model
#@markdown ___
#@markdown Removed the encoder type section because we all know we ain't touching that.
encoder_type = "whisper"
whisper_model = "openai/whisper-base"
wavlm_model = "microsoft/wavlm-base-plus"
freeze_encoder = True #@param {type:"boolean"}
unfreeze_last_n_layers = 2 #@param {type: "number"}
#@markdown <font size="-1.5"> Unfreeze last 2 layers to adapt to speaker's mic/accent

#@markdown # Dataset size
#@markdown ___
#@markdown small = under 30 min\
#@markdown medium = 30 min to 4 hrs\
#@markdown high = 4+ hrs
data_size = "medium" #@param ["small", "medium", "high"]
#@markdown # Training
#@markdown ___
#@markdown <font size="-1.5"> I'm honestly not sure how the vram usage is calculated, so I'm just leaving it at the config defaults
batch_size = 8 #@param {type: "integer"}
num_workers = 4 #@param {type: "integer"}
learning_rate = 1e-4 #@param {type: "number"}
#@markdown <font size="-1.5"> It's recommended to keep augmentation on during initial training, disable and resume training when accuracy/loss stagnates.
data_aug = True #@param {type: "boolean"}
#@markdown <font size="-1.5"> learning rate decay gamma every val_check_interval
lr_decay_gamma = 0.9 #@param {type: "number"}
#@markdown <font size="-1.5"> prevent `overconfident` labels
label_smoothing = 0.1 #@param {type: "number"}
#markdown <font size="-1.5"> path to logs folder for tensorboard
#log_dir = "/content/drive/MyDrive/WFL/Model/logs" #@param {type: "string"}
#markdown <font size="-1.5"> path to checkpoints folder
#@markdown # Finetuning
#@markdown ___
#@markdown <font size="-1.5"> enabling WLF finetuning
finetune_enable = False #@param {type:"boolean"}
#@markdown <font size="-1.5"> path to finetune model
finetune_model_path = "" #@param {type: "string"}
#@markdown Do note that the number of languages have to match up or it'll throw an error.
#@markdown # Saving
#@markdown ___
#@markdown <font size="-1.5"> path to save folder
save_dir = "/content/drive/MyDrive/testwfl" #@param {type: "string"}
#@markdown <font size="-1.5"> training stop point
max_epochs = 1000 # @param {"type":"slider","min":500,"max":2000,"step":100}
#@markdown <font size="-1.5"> validation/saving interval
val_check_interval = 2 # @param {"type":"slider","min":1,"max":5,"step":1}

if not save_dir:
    raise ValueError("save_dir is not set, please set a saving directory")
import re
import os
import yaml

log_dir = os.path.join(save_dir, "logs")

with open(f"/content/WFL-ASR/configs/{data_size}.yaml", "r") as config:
    wfl_config = yaml.safe_load(config)
#data
wfl_config["data"]["data_dir"] = data_dir
#model
wfl_config["model"]["encoder_type"] = encoder_type
wfl_config["model"]["whisper_model"] = whisper_model
wfl_config["model"]["wavlm_model"] = wavlm_model
wfl_config["model"]["freeze_encoder"] = freeze_encoder
#training
wfl_config["training"]["batch_size"] = batch_size
wfl_config["training"]["num_workers"] = num_workers
wfl_config["training"]["learning_rate"] = learning_rate
wfl_config["training"]["lr_decay_gamma"] = lr_decay_gamma
wfl_config["training"]["label_smoothing"] = label_smoothing
wfl_config["training"]["max_epochs"] = max_epochs
wfl_config["training"]["check_val_every_n_epoch"] = val_check_interval
wfl_config["training"]["log_dir"] = log_dir
wfl_config["augmentation"]["enable"] = data_aug
#finetuning
wfl_config["finetune"]["enabled"] = finetune_enable
wfl_config["finetune"]["checkpoint_path"] = finetune_model_path
#output
wfl_config["output"]["save_dir"] = save_dir

with open("/content/WFL-ASR/config.yaml", "w") as config:
    yaml.dump(wfl_config, config)
````
